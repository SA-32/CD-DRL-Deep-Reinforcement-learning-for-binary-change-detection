```python
import os
import numpy as np
import torch
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader
from torchvision import transforms

from dataset import LEVIRCDPatchDataset
from env import PixelEnv, NeighborhoodAggregator
from model import PixelDRLMG


@torch.no_grad()
def predict_episode(
    model,
    env,
    pre_change_image,
    post_change_image,
    gt,
    t_max,
    device
):
    """
    Run the trained PixelDRLMG model in inference mode.

    Returns:
        final_mask
        mask_history
    """

    mask = env.init_mask(pre_change_image)

    mask_history = [mask.clone()]

    for t in range(t_max):

        x_t = torch.cat(
            [
                pre_change_image,
                post_change_image,
                mask
            ],
            dim=1
        )

        # IMPORTANT:
        # greedy=True makes prediction deterministic
        action, _, _, _, _ = model.act(
            x_t,
            t, greedy = True
        )

        new_mask_2d = env.step(
            mask.squeeze(1),
            action
        )

        mask = new_mask_2d.unsqueeze(1)

        mask_history.append(mask.clone())

    return mask, mask_history
```

```python
def plot_prediction(
    pre_image,
    post_image,
    gt,
    pred,
    mask_history,
    sample_idx,
    save_dir="predictions"
):

    os.makedirs(save_dir, exist_ok=True)

    # ------------------------------------------------
    # ImageNet denormalization
    # ------------------------------------------------

    mean = torch.tensor(
        [0.485, 0.456, 0.406],
        device=pre_image.device
    ).view(1, 3, 1, 1)

    std = torch.tensor(
        [0.229, 0.224, 0.225],
        device=pre_image.device
    ).view(1, 3, 1, 1)

    # Denormalize
    pre_denorm = pre_image * std + mean
    post_denorm = post_image * std + mean

    # ------------------------------------------------
    # Convert tensors to numpy
    # ------------------------------------------------

    pre = pre_denorm[0].detach().cpu().numpy()
    post = post_denorm[0].detach().cpu().numpy()

    gt_np = gt[0, 0].detach().cpu().numpy()

    # If prediction is logits, apply sigmoid
    pred_prob = torch.sigmoid(pred[0, 0]).detach().cpu().numpy()

    # ------------------------------------------------
    # CHW -> HWC
    # ------------------------------------------------

    if pre.shape[0] == 3:
        pre = np.transpose(pre, (1, 2, 0))

    if post.shape[0] == 3:
        post = np.transpose(post, (1, 2, 0))

    # ------------------------------------------------
    # Clip after denormalization
    # ------------------------------------------------

    pre = np.clip(pre, 0, 1)
    post = np.clip(post, 0, 1)

    # ------------------------------------------------
    # Prediction
    # ------------------------------------------------

    pred_binary = (pred_prob > 0.5).astype(np.float32)

    # ------------------------------------------------
    # Error map
    # ------------------------------------------------

    error = np.abs(gt_np - pred_binary)

    # ------------------------------------------------
    # Rollout
    # ------------------------------------------------

    n_steps = len(mask_history)

    # Handle case where there are fewer than 5 steps
    if n_steps >= 5:
        selected_steps = np.linspace(
            0,
            n_steps - 1,
            5,
            dtype=int
        )
    else:
        selected_steps = np.arange(n_steps)

    # ------------------------------------------------
    # Plot
    # ------------------------------------------------

    fig, axes = plt.subplots(
        2,
        5,
        figsize=(18, 8)
    )

    # =================================================
    # ROW 1
    # =================================================

    axes[0, 0].imshow(pre)
    axes[0, 0].set_title("Pre-change")
    axes[0, 0].axis("off")

    axes[0, 1].imshow(post)
    axes[0, 1].set_title("Post-change")
    axes[0, 1].axis("off")

    axes[0, 2].imshow(
        gt_np,
        cmap="gray",
        vmin=0,
        vmax=1
    )
    axes[0, 2].set_title("Ground Truth")
    axes[0, 2].axis("off")

    axes[0, 3].imshow(
        pred_binary,
        cmap="gray",
        vmin=0,
        vmax=1
    )
    axes[0, 3].set_title("Prediction")
    axes[0, 3].axis("off")

    axes[0, 4].imshow(
        error,
        cmap="hot",
        vmin=0,
        vmax=1
    )
    axes[0, 4].set_title("Error")
    axes[0, 4].axis("off")

    # =================================================
    # ROW 2 — ROLLOUT
    # =================================================

    # Hide unused rollout axes
    for j in range(5):
        axes[1, j].axis("off")

    for j, step in enumerate(selected_steps):

        mask = (
            mask_history[step][0, 0]
            .detach()
            .cpu()
            .numpy()
        )

        axes[1, j].imshow(
            mask,
            cmap="gray",
            vmin=0,
            vmax=1
        )

        axes[1, j].set_title(
            f"Rollout t={step}"
        )

        axes[1, j].axis("off")

    # ------------------------------------------------
    # Overall title
    # ------------------------------------------------

    plt.suptitle(
        f"PixelDRLMG Prediction - Sample {sample_idx}",
        fontsize=16
    )

    plt.tight_layout()

    # ------------------------------------------------
    # Save
    # ------------------------------------------------

    save_path = os.path.join(
        save_dir,
        f"sample_{sample_idx}.png"
    )

    plt.savefig(
        save_path,
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    print(f"Saved: {save_path}")
```

```python
def inference():

    # ------------------------------------------------
    # Device
    # ------------------------------------------------

    device = torch.device(
        "cuda" if torch.cuda.is_available() else "cpu"
    )

    print("Device:", device)

    # ------------------------------------------------
    # Dataset
    # ------------------------------------------------

    transform = transforms.Compose([
        transforms.Resize((256, 256)),
        transforms.ToTensor()
    ])

    root_dir = "../../levir_split/test"

    test_ds = LEVIRCDPatchDataset(
        root_dir,
        mask_transform=transform
    )

    test_loader = DataLoader(
        test_ds,
        batch_size=8,
        shuffle=False,
        num_workers=2
    )

    # ------------------------------------------------
    # Model
    # ------------------------------------------------

    model = PixelDRLMG().to(device)

    aggregator = NeighborhoodAggregator().to(device)

    # ------------------------------------------------
    # Load checkpoint
    # ------------------------------------------------

    checkpoint = torch.load(
        "pixeldrl_mg_epoch100.pt",
        map_location=device
    )

    model.load_state_dict(
        checkpoint["model"]
    )

    aggregator.load_state_dict(
        checkpoint["aggregator"]
    )

    print(
        f"Loaded checkpoint from epoch "
        f"{checkpoint['epoch']}"
    )

    # ------------------------------------------------
    # Evaluation mode
    # ------------------------------------------------

    model.train()
    aggregator.eval()

    # ------------------------------------------------
    # Environment
    # ------------------------------------------------

    env = PixelEnv(device)

    # ------------------------------------------------
    # Predictions
    # ------------------------------------------------
    for name, m in model.named_modules():
        if isinstance(m, (torch.nn.BatchNorm1d, torch.nn.BatchNorm2d)):
            print(name, "num_batches_tracked:", m.num_batches_tracked.item(),
                  "running_var mean:", m.running_var.mean().item(),
                  "running_var min:", m.running_var.min().item())

    for sample_idx, batch in enumerate(test_loader):

        pre_change_image, post_change_image, gt = batch

        pre_change_image = pre_change_image.to(device)
        post_change_image = post_change_image.to(device)
        gt = gt.to(device)

        # --------------------------------------------
        # Prediction
        # --------------------------------------------

        final_mask, mask_history = predict_episode(
            model=model,
            env=env,
            pre_change_image=pre_change_image,
            post_change_image=post_change_image,
            gt=gt,
            t_max=10,
            device=device
        )

        # --------------------------------------------
        # Plot
        # --------------------------------------------

        plot_prediction(
            pre_image=pre_change_image,
            post_image=post_change_image,
            gt=gt,
            pred=final_mask,
            mask_history=mask_history,
            sample_idx=sample_idx
        )

        # --------------------------------------------
        # Only plot first few samples
        # --------------------------------------------

        if sample_idx >= 4:
            break
```

```python
if __name__ == "__main__":
    inference()
```
